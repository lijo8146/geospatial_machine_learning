## Before you run

`H3HexagonLayer` displays existing H3 cells. `HexagonLayer` bins raw points into a different grid; using it here would reaggregate the data. The filename is retained from the starter for continuity. Browser rendering needs internet for JavaScript assets; the CSV analysis remains local.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "course.py").exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from course import OUT, points
```

## Render count and score explicitly


```python
import pydeck as pdk
from course import embeddings,evaluate,h3_summary
frame=points(2000)
x=embeddings(frame); model,*_=evaluate(frame,x)
frame['probability']=model.predict_proba(x)[:,1]
summary=h3_summary(frame,7)
summary['color']=[[int(255*p),80,int(255*(1-p)),190] for p in summary.mean_probability]
layer=pdk.Layer('H3HexagonLayer',summary,get_hexagon='hex',get_fill_color='color',pickable=True,stroked=True,filled=True)
deck=pdk.Deck(layers=[layer],initial_view_state=pdk.ViewState(latitude=39.98,longitude=-105.22,zoom=10),map_style=None,
    tooltip={'text':'H3: {hex}\nSynthetic mean class-1 score: {mean_probability}\nSample count: {count}'})
html=OUT/'synthetic_h3_map.html'
deck.to_html(str(html),open_browser=False)
print('Saved',html)
deck
```
